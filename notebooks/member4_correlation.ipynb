***3. Select 3 numerical variables***

```python
selected_columns = [
    "math score",
    "reading score",
    "writing score"
]

performance_df = df[selected_columns]

performance_df.head()
```

***4. Correlation Matrix***

```python
correlation_matrix = performance_df.corr()

correlation_matrix
```

***5. Correlation Heatmap***

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Student Performance")
plt.xlabel("Subjects")
plt.ylabel("Subjects")
plt.tight_layout()

plt.show()
```

***Correlation Analysis***

The correlation matrix measures the strength and direction of the relationships
between math, reading, and writing scores.

A positive correlation indicates that higher scores in one subject tend to be
associated with higher scores in another subject. The heatmap provides a visual
representation of these relationships, where the correlation values range from
-1 to +1.

The diagonal values are 1 because each variable is perfectly correlated with
itself.
