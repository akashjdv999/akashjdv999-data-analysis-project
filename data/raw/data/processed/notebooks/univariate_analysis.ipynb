```python
print("Mean:", df["math score"].mean())
print("Median:", df["math score"].median())
print("Mode:", df["math score"].mode()[0])
print("Minimum:", df["math score"].min())
print("Maximum:", df["math score"].max())
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["math score"], bins=10, edgecolor="black")

plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.title("Distribution of Math Scores")

plt.show()
```

```python
plt.figure(figsize=(8, 4))

plt.boxplot(df["math score"], vert=False)

plt.xlabel("Math Score")
plt.title("Box Plot of Math Scores")

plt.show()
```

```python
gender_counts = df["gender"].value_counts()

print(gender_counts)
```

```python
gender_counts.plot(kind="bar")

plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.title("Gender Distribution")

plt.show()
```

```python
categorical_columns = [
    "gender",
    "race/ethnicity",
    "parental level of education",
    "lunch",
    "test preparation course"
]

for col in categorical_columns:
    print("\n", col)
    print(df[col].value_counts())
    
    df[col].value_counts().plot(kind="bar", figsize=(8, 4))
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.title(f"{col} Distribution")
    plt.xticks(rotation=45)
    plt.show()
```
