```python
# ============================================================
# STUDENT PERFORMANCE DATASET
# MEMBER 3 - BIVARIATE ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

df = pd.read_csv(r"C:\Users\shaik\Downloads\preprocessing\akashjdv999-data-analysis-project\data\raw\student_performance.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())

# ------------------------------------------------------------
# 2. BASIC INFORMATION
# ------------------------------------------------------------

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

# ------------------------------------------------------------
# 3. NUMERICAL VS NUMERICAL
# ------------------------------------------------------------

print("\n" + "="*60)
print("NUMERICAL VS NUMERICAL ANALYSIS")
print("="*60)

# Correlation between scores
score_columns = [
    "math score",
    "reading score",
    "writing score"
]

print("\nCorrelation Matrix:")
print(df[score_columns].corr())

# ------------------------------------------------------------
# 4. MATH VS READING
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="math score",
    y="reading score"
)
plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.show()

print(
    "Math vs Reading Correlation:",
    round(df["math score"].corr(df["reading score"]), 3)
)

# ------------------------------------------------------------
# 5. MATH VS WRITING
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="math score",
    y="writing score"
)
plt.title("Math Score vs Writing Score")
plt.xlabel("Math Score")
plt.ylabel("Writing Score")
plt.show()

print(
    "Math vs Writing Correlation:",
    round(df["math score"].corr(df["writing score"]), 3)
)

# ------------------------------------------------------------
# 6. READING VS WRITING
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="reading score",
    y="writing score"
)
plt.title("Reading Score vs Writing Score")
plt.xlabel("Reading Score")
plt.ylabel("Writing Score")
plt.show()

print(
    "Reading vs Writing Correlation:",
    round(df["reading score"].corr(df["writing score"]), 3)
)

# ------------------------------------------------------------
# 7. GENDER VS SCORES
# ------------------------------------------------------------

print("\n" + "="*60)
print("GENDER VS SCORES")
print("="*60)

print("\nAverage Scores by Gender:")
print(
    df.groupby("gender")[
        ["math score", "reading score", "writing score"]
    ].mean()
)

# Gender vs Math
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="gender",
    y="math score"
)
plt.title("Math Score Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Math Score")
plt.show()

# Gender vs Reading
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="gender",
    y="reading score"
)
plt.title("Reading Score Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Reading Score")
plt.show()

# Gender vs Writing
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="gender",
    y="writing score"
)
plt.title("Writing Score Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Writing Score")
plt.show()

# ------------------------------------------------------------
# 8. TEST PREPARATION VS SCORES
# ------------------------------------------------------------

print("\n" + "="*60)
print("TEST PREPARATION VS SCORES")
print("="*60)

print("\nAverage Scores by Test Preparation:")
print(
    df.groupby("test preparation course")[
        ["math score", "reading score", "writing score"]
    ].mean()
)

# Math
plt.figure(figsize=(9, 5))
sns.boxplot(
    data=df,
    x="test preparation course",
    y="math score"
)
plt.title("Math Score by Test Preparation Course")
plt.xlabel("Test Preparation Course")
plt.ylabel("Math Score")
plt.show()

# Reading
plt.figure(figsize=(9, 5))
sns.boxplot(
    data=df,
    x="test preparation course",
    y="reading score"
)
plt.title("Reading Score by Test Preparation Course")
plt.xlabel("Test Preparation Course")
plt.ylabel("Reading Score")
plt.show()

# Writing
plt.figure(figsize=(9, 5))
sns.boxplot(
    data=df,
    x="test preparation course",
    y="writing score"
)
plt.title("Writing Score by Test Preparation Course")
plt.xlabel("Test Preparation Course")
plt.ylabel("Writing Score")
plt.show()

# ------------------------------------------------------------
# 9. LUNCH VS SCORES
# ------------------------------------------------------------

print("\n" + "="*60)
print("LUNCH VS SCORES")
print("="*60)

print("\nAverage Scores by Lunch Type:")
print(
    df.groupby("lunch")[
        ["math score", "reading score", "writing score"]
    ].mean()
)

# Math
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="lunch",
    y="math score"
)
plt.title("Math Score by Lunch Type")
plt.xlabel("Lunch Type")
plt.ylabel("Math Score")
plt.show()

# Reading
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="lunch",
    y="reading score"
)
plt.title("Reading Score by Lunch Type")
plt.xlabel("Lunch Type")
plt.ylabel("Reading Score")
plt.show()

# Writing
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="lunch",
    y="writing score"
)
plt.title("Writing Score by Lunch Type")
plt.xlabel("Lunch Type")
plt.ylabel("Writing Score")
plt.show()

# ------------------------------------------------------------
# 10. PARENTAL EDUCATION VS MATH SCORE
# ------------------------------------------------------------

print("\n" + "="*60)
print("PARENTAL EDUCATION VS MATH SCORE")
print("="*60)

print("\nAverage Math Score by Parental Education:")
print(
    df.groupby("parental level of education")["math score"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))
sns.boxplot(
    data=df,
    x="parental level of education",
    y="math score"
)
plt.title("Math Score by Parental Level of Education")
plt.xlabel("Parental Level of Education")
plt.ylabel("Math Score")
plt.xticks(rotation=30)
plt.show()

# ------------------------------------------------------------
# 11. RACE/ETHNICITY VS MATH SCORE
# ------------------------------------------------------------

print("\n" + "="*60)
print("RACE/ETHNICITY VS MATH SCORE")
print("="*60)

print("\nAverage Math Score by Race/Ethnicity:")
print(
    df.groupby("race/ethnicity")["math score"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="race/ethnicity",
    y="math score"
)
plt.title("Math Score by Race/Ethnicity")
plt.xlabel("Race/Ethnicity")
plt.ylabel("Math Score")
plt.show()

# ------------------------------------------------------------
# 12. CATEGORICAL VS CATEGORICAL
# ------------------------------------------------------------

print("\n" + "="*60)
print("CATEGORICAL VS CATEGORICAL ANALYSIS")
print("="*60)

# Gender vs Test Preparation
print("\nGender vs Test Preparation:")
gender_testprep = pd.crosstab(
    df["gender"],
    df["test preparation course"]
)
print(gender_testprep)

gender_testprep.plot(
    kind="bar",
    figsize=(8, 5)
)
plt.title("Gender vs Test Preparation Course")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.show()

# Gender vs Lunch
print("\nGender vs Lunch:")
gender_lunch = pd.crosstab(
    df["gender"],
    df["lunch"]
)
print(gender_lunch)

gender_lunch.plot(
    kind="bar",
    figsize=(8, 5)
)
plt.title("Gender vs Lunch Type")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.show()

# Parental Education vs Test Preparation
print("\nParental Education vs Test Preparation:")
parent_testprep = pd.crosstab(
    df["parental level of education"],
    df["test preparation course"]
)
print(parent_testprep)

parent_testprep.plot(
    kind="bar",
    figsize=(10, 6)
)
plt.title("Parental Education vs Test Preparation Course")
plt.xlabel("Parental Level of Education")
plt.ylabel("Number of Students")
plt.xticks(rotation=30)
plt.show()

# Race/Ethnicity vs Lunch
print("\nRace/Ethnicity vs Lunch:")
race_lunch = pd.crosstab(
    df["race/ethnicity"],
    df["lunch"]
)
print(race_lunch)

race_lunch.plot(
    kind="bar",
    figsize=(8, 5)
)
plt.title("Race/Ethnicity vs Lunch Type")
plt.xlabel("Race/Ethnicity")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.show()

# ------------------------------------------------------------
# 13. FINAL CORRELATION SUMMARY
# ------------------------------------------------------------

print("\n" + "="*60)
print("CORRELATION SUMMARY")
print("="*60)

correlation_summary = pd.DataFrame({
    "Relationship": [
        "Math vs Reading",
        "Math vs Writing",
        "Reading vs Writing"
    ],
    "Correlation": [
        df["math score"].corr(df["reading score"]),
        df["math score"].corr(df["writing score"]),
        df["reading score"].corr(df["writing score"])
    ]
})

print(correlation_summary)

# ------------------------------------------------------------
# 14. END
# ------------------------------------------------------------

print("\n" + "="*60)
print("BIVARIATE ANALYSIS COMPLETED SUCCESSFULLY")
print("="*60)
```
